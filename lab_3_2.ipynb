```python
# Задание task_03_02_07.
# Выполнила: Михалева Полина Вячеславовна
# Группа: ЦИБ-251

def foo(low, high):
    """
    Алгоритм: бинарный поиск (угадывание числа)
    Сложность: O(log N)

    Параметры:
    - low (int): нижняя граница;
    - high (int): верхняя граница.
    """
    угадываю = True
    while угадываю:
        угадай = (low + high) // 2
        print("Загаданное число {0}?".format(угадай))
        указатель = input(
            "Введите '+', если ваше число меньше.\n"
            "Введите '-', если ваше число больше.\n"
            "Введите '=', если я угадал.\n"
        ).lower()

        if указатель == '+':
            high = угадай - 1
        elif указатель == '-':
            low = угадай + 1
        elif указатель == '=':
            угадываю = False
        else:
            print("Введите '+', '-' или '='.")

    return угадай


#low, high = 0, 100
#print("Пожалуйста, загадайте число от {0} до {1}!".format(low, high))
#guess = foo(low, high)
#print("Игра окончена, вы угадали число: {0}.".format(guess))


```
