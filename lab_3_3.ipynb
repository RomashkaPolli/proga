```python
# Задание task_03_03_06.
# Выполнила: Михалева Полина Вячеславовна
# Группа: ЦИБ-251

"""
Ошибки (номера строк через пробел, данная строка - №2): 19 22 23
"""

def уровень_безработицы(безработные, занятые):
    """Вернуть уровень безработицы (УБ) в долях 1.

    Расчет по формуле: УБ = Безработные / (Занятые + Безработные).
    """
    if безработные < 0 or занятые < 0:
        raise ValueError("Количество людей не может быть отрицательным.")

    if безработные == 0 and занятые == 0:
        raise ZeroDivisionError("Общее количество людей не может быть равно нулю.")

    return безработные / (безработные + занятые)


try:
    безработные = int(input("Введите количество безработных (чел.): "))
    занятые = int(input("Введите количество занятых (чел.): "))

    try:
        rate = уровень_безработицы(безработные, занятые)
        print("Уровень безработицы = {:.1%}".format(rate))
    except ValueError as ve:
        print("Ошибка ввода:", ve)
    except ZeroDivisionError:
        print("Ошибка ввода: Общее количество людей не может быть равно нулю.")

except ValueError:
    print("Ошибка ввода: Пожалуйста, вводите только целые числа для количества людей.")
```
