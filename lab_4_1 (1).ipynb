```python
"""
Лабораторная работа № 4.1.
Работу выполнила: Михалева Полина Вячеславовна, ЦИБ-251
Вариант 17: Автосалон
Базовый класс 1: Vehicle (цена)
Базовый класс 2: Buyer (id)
"""

class Vehicle:
    """Базовый класс 1: Транспортное средство"""
    def __init__(self, model: str, price: float):
        if not isinstance(model, str) or not model.strip():
            raise ValueError("Модель автомобиля должна быть непустой строкой.")
        if price < 0:
            raise ValueError("Цена автомобиля не может быть отрицательной.")

        self.model = model
        self.price = price

    def display_info(self) -> None:
        print(f"Автомобиль: {self.model} | Цена: {self.price:,.2f} руб.")

    def __str__(self) -> str:
        return f" Автомобиль: {self.model} |  Цена: {self.price:,.2f} руб."


class Buyer:
    """Базовый класс 2: Покупатель"""
    def __init__(self, buyer_id: int, name: str):
        if buyer_id <= 0:
            raise ValueError("ID покупателя должен быть положительным числом.")
        if not isinstance(name, str) or not name.strip():
            raise ValueError("Имя покупателя должно быть непустой строкой.")

        self.buyer_id = buyer_id
        self.name = name

    def display_info(self) -> None:
        print(f"Покупатель: {self.name} | ID: {self.buyer_id}")

    def __str__(self) -> str:
        return f" Покупатель: {self.name} |  ID: {self.buyer_id}"


# Блок демонстрации с обработкой исключений
if __name__ == "__main__":
    print("=== 1. Корректное создание объектов ===")
    try:
        car1 = Vehicle(model="Toyota Camry", price=3500000.0)
        buyer1 = Buyer(buyer_id=101, name="Иван Иванов")
        print("Объекты успешно созданы:")
        print(car1)
        print(buyer1)
        buyer2 = Buyer(buyer_id=102, name = "Аня")
        print(buyer2)
    except ValueError as e:
        print(f"Ошибка при создании: {e}")

    print("\n=== 2. Демонстрация обработки ошибок (try...except) ===")

    # Попытка создать автомобиль с отрицательной ценой
    try:
        print("Пытаемся создать автомобиль с отрицательной ценой...")
        bad_car = Vehicle(model="Lada", price=-500000.0)
    except ValueError as e:
        print(f"Перехвачена ошибка ValueError: {e}")

    # Попытка создать покупателя с пустым именем
    try:
        print("Пытаемся создать покупателя с пустым именем...")
        bad_buyer = Buyer(buyer_id=102, name="   ")
    except ValueError as e:
        print(f"Перехвачена ошибка ValueError: {e}")
```
