```python
"""
Лабораторная работа № 4.1.
Работу выполнила: Михалева Полина Вячеславовна, ЦИБ-251
Вариант 17: Автосалон
Базовый класс 1: Vehicle (цена)
Базовый класс 2: Buyer (id)
"""

class Vehicle:
    """
    Класс, представляющий автомобиль (сущность) в автосалоне.

    Атрибуты:
        model (str): Модель автомобиля.
        price (float): Цена автомобиля в рублях.
    """

    def __init__(self, model: str, price: float):
        self.model = model
        self.price = price

    def display_info(self) -> None:
        print(f"Автомобиль: {self.model} | Цена: {self.price:,.2f} руб.")

    def __str__(self) -> str:
        return f"Автомобиль: {self.model} | Цена: {self.price:,.2f} руб."


class Buyer:
    """
    Класс, представляющий покупателя (субъект) в автосалоне.

    Атрибуты:
        buyer_id (int): Уникальный идентификатор покупателя.
        name (str): Имя покупателя.
    """

    def __init__(self, buyer_id: int, name: str):
        self.buyer_id = buyer_id
        self.name = name

    def display_info(self) -> None:
        print(f"Покупатель: {self.name} | ID: {self.buyer_id}")

    def __str__(self) -> str:
        return f"Покупатель: {self.name} | ID: {self.buyer_id}"


# Блок демонстрации работы кода
if __name__ == "__main__":
    print("=== Создание объектов класса Vehicle ===")
    # Создаем 3 объекта первого класса с разными данными
    car1 = Vehicle(model="Toyota Camry", price=3500000.0)
    car2 = Vehicle(model="BMW X5", price=8750000.50)
    car3 = Vehicle(model="Lada Vesta NG", price=1450000.0)

    print("\n=== Создание объектов класса Buyer ===")
    # Создаем 2 объекта второго класса
    buyer1 = Buyer(buyer_id=1001, name="Иван Иванов")
    buyer2 = Buyer(buyer_id=1002, name="Анна Петрова")

    print("\n=== Демонстрация работы метода __str__ (Vibe Code) ===")
    # Выводим все объекты в консоль через print()
    print(car1)
    print(car2)
    print(car3)
    print(buyer1)
    print(buyer2)

    print("\n=== Демонстрация работы метода display_info() ===")
    car1.display_info()
    buyer1.display_info()
```
